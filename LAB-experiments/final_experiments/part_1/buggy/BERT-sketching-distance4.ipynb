```python
import math
import bisect
import random
import numpy as np

class KLLSketch:
    """
    A more "canonical" KLL sketch for scalar data, with:
      - Hierarchical capacities (small at lower levels, larger at higher levels)
      - Randomized compaction to reduce bias
      - Mergeability
      - Bottom-level sampler (capacity=2)
    Each item is stored as (value, weight).
    """

    def __init__(self, max_capacity=200, max_levels=10):
        """
        :param max_capacity: The largest capacity we allow at higher levels.
        :param max_levels:   The maximum number of levels (logarithmic in data size).
        """
        self.max_capacity = max_capacity
        # You can tweak how capacity grows by level. Here is one simple approach:
        # Level 0: capacity=2 (sampler), then capacity doubles up to a max.
        # Example: [2, 4, 8, 16, 32, ..., up to max_capacity]
        self.level_capacities = [2]
        while len(self.level_capacities) < max_levels:
            next_cap = min(self.level_capacities[-1] * 2, self.max_capacity)
            self.level_capacities.append(next_cap)

        # Each level is a list of (value, weight)
        self.levels = [[] for _ in range(max_levels)]
        # Track how many levels we are actively using
        self.active_levels = 1  # start with just level 0 in use
        self.n = 0  # total number of items seen (sum of original weights)

    def update(self, value):
        """
        Insert a single scalar value. It starts at level 0 with weight=1.
        If level 0 overflows, we compact it, possibly cascading upwards.
        """
        # Bottom-level insertion (weight=1)
        self.n += 1
        self.levels[0].append((value, 1))

        # Check if bottom-level overflows
        if len(self.levels[0]) > self.level_capacities[0]:
            self._compact_level(0)

    def merge(self, other_sketch):
        """
        Merge another KLLSketch into this one. We simply concatenate
        each level and compact if needed. This is the standard merging approach.
        """
        # Ensure we have the same number of levels
        for lvl in range(len(self.levels)):
            # Extend our level with the other's items
            self.levels[lvl].extend(other_sketch.levels[lvl])

        # Merge total counts
        self.n += other_sketch.n

        # Now compact from bottom to top if needed
        for lvl in range(self.active_levels):
            cap = self.level_capacities[lvl]
            if len(self.levels[lvl]) > cap:
                self._compact_level(lvl)

    def _compact_level(self, level_idx):
        """
        Compacts (halves) the items at level_idx if they exceed capacity,
        pushing the retained items up to the next level with doubled weight.
        If that next level also overflows, it cascades upward, etc.
        """
        level_buffer = self.levels[level_idx]
        capacity = self.level_capacities[level_idx]

        # If this level is still within capacity, nothing to do
        if len(level_buffer) <= capacity:
            return

        # Sort by value
        level_buffer.sort(key=lambda x: x[0])  # sort by .value

        # Randomly choose whether to discard even or odd indices
        discard_even = (random.random() < 0.5)

        # We'll keep half of them (rounded down).
        # The ones we keep get weight * 2 and move to next level.
        kept = []
        for i, (val, w) in enumerate(level_buffer):
            # i%2 == 0 => "even index"
            # If discard_even is True, we discard even indices => keep odd.
            # If discard_even is False, we discard odd indices => keep even.
            if discard_even:
                if i % 2 == 1:
                    kept.append((val, w * 2))
            else:
                if i % 2 == 0:
                    kept.append((val, w * 2))

        # Clear the current level
        self.levels[level_idx] = []

        # If we don't have a next level active, activate it
        if level_idx + 1 >= self.active_levels:
            self.active_levels += 1

        # Push these kept items to the next level
        self.levels[level_idx + 1].extend(kept)

        # Now check if the next level is over capacity (cascade)
        if len(self.levels[level_idx + 1]) > self.level_capacities[level_idx + 1]:
            self._compact_level(level_idx + 1)

    def _sorted_items(self):
        """
        Return a single sorted list of (value, weight) from all active levels.
        """
        merged = []
        for lvl in range(self.active_levels):
            merged.extend(self.levels[lvl])
        merged.sort(key=lambda x: x[0])
        return merged

    def rank(self, value):
        """
        Return the approximate rank of 'value' (i.e., how many are <= value).
        Uses the item weights to compute the rank.
        """
        items = self._sorted_items()

        # We can do a binary search approach, but let's keep it simple:
        # 1) find the insertion point (bisect_right by value)
        # 2) sum up weights of all items <= that position
        # Note that we cannot do a direct bisect on items unless we only store values,
        # but we can do a manual approach or we can do a bisect with a custom key.
        # For clarity, let's just iterate once for now, but you could optimize further.
        # We'll do a linear pass or a custom binary search. For demonstration, linear is fine.
        # For large data, you'd store partial sums or do a more advanced binary search.

        rank_sum = 0
        for (v, w) in items:
            if v <= value:
                rank_sum += w
            else:
                break
        return rank_sum

    def cdf(self, value):
        """
        Return the fraction of items <= value. That is rank(value) / total_count.
        """
        if self.n == 0:
            return 0.0
        return self.rank(value) / float(self.n)

class KLLDimensionReducer:
    """
    This class maintains one KLLSketch per dimension, transforming each
    embedding dimension to its rank-based percentile in [0,1].
    Optionally, it can then apply a second reduction step (e.g., PCA)
    to map from D dimensions down to fewer dims.
    """

    def __init__(self, max_capacity=200, max_levels=10, reduce_dims=None):
        """
        :param max_capacity: Largest capacity for top levels in each KLLSketch.
        :param max_levels:   Max number of levels in each KLLSketch.
        :param reduce_dims:  If not None, the dimensionality to reduce to (e.g. 10, 50, etc.).
                             If None, no secondary reduction is applied.
        """
        self.max_capacity = max_capacity
        self.max_levels = max_levels
        self.reduce_dims = reduce_dims

        self.sketches = []        # One KLLSketch per dimension
        self.fitted = False
        self.reducer_model = None # e.g. a PCA model if reduce_dims is set

    def fit(self, baseline_embs):
        """
        Build one KLLSketch per dimension, then optionally learn a secondary
        transformation (e.g., PCA) on the rank-based vectors.
        :param baseline_embs: numpy array of shape (num_samples, D)
        """
        num_dims = baseline_embs.shape[1]
        self.sketches = [
            KLLSketch(max_capacity=self.max_capacity, max_levels=self.max_levels)
            for _ in range(num_dims)
        ]

        # Update each dimension's sketch with the baseline data
        for row in baseline_embs:
            for d in range(num_dims):
                self.sketches[d].update(row[d])

        # 2) If reduce_dims is specified, learn PCA on the entire rank-based data
        if self.reduce_dims is not None and self.reduce_dims < num_dims:
            # Transform the entire baseline to their rank-based representation
            rank_vectors = [self._ranks_for_one(x) for x in baseline_embs]
            rank_vectors = np.array(rank_vectors)  # shape: (num_samples, D)

            # Fit PCA (or another chosen method)
            self.reducer_model = PCA(n_components=self.reduce_dims)
            self.reducer_model.fit(rank_vectors)

        self.fitted = True

    def _ranks_for_one(self, emb):
        """
        Utility: convert a single embedding (D-dim) to a rank-based vector in [0,1]^D.
        """
        return [sk.cdf(val) for sk, val in zip(self.sketches, emb)]

    def transform_one(self, emb):
        """
        Map a single embedding to rank space [0,1]^D, then optionally apply PCA.
        """
        if not self.fitted:
            raise ValueError("KLLDimensionReducer not fitted yet!")

        # 1) Rank-based vector in [0,1]^D
        rank_vec = np.array(self._ranks_for_one(emb))

        # 2) Optional secondary reduction
        if self.reducer_model is not None:
            rank_vec = self.reducer_model.transform(rank_vec.reshape(1, -1))[0]

        return rank_vec

    def transform_batch(self, embs):
        """
        Convert a batch of embeddings (batch_size, D) into rank-based space
        and optionally reduce further. Returns shape: (batch_size, new_dim).
        """
        if not self.fitted:
            raise ValueError("KLLDimensionReducer not fitted yet!")

        # Create rank-based representation
        rank_matrix = np.array([self._ranks_for_one(e) for e in embs])  # (B, D)

        # If we have a secondary reducer, apply it
        if self.reducer_model is not None:
            rank_matrix = self.reducer_model.transform(rank_matrix)

        return rank_matrix

    def merge(self, other_reducer):
        """
        Merge this dimension reducer with another one of the same dimension.
        Only merges the KLL sketches (rank transformations). The PCA model
        can't be trivially merged, so you'd typically re-fit it after merging
        if you want a consistent global reduction.
        """
        if len(self.sketches) != len(other_reducer.sketches):
            raise ValueError("Dimension mismatch in KLLDimensionReducer merge.")

        # Merge each dimension's KLLSketch
        for d in range(len(self.sketches)):
            self.sketches[d].merge(other_reducer.sketches[d])

        # We do NOT merge the PCA model automatically. If a user
        # wants a single global PCA after merging, they'd call fit again
        # or adopt an incremental PCA approach.
```

```python
import os
import random
import torch
import time
import numpy as np

from matplotlib import pyplot as plt
from transformers import AutoTokenizer, AutoModel
from datasets import load_dataset
from tqdm import tqdm
from collections import defaultdict
import math
from sklearn.decomposition import PCA




###############################################################################
# 1) Utility Functions
###############################################################################
def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

def batch_generator(data, batch_size=32):
    for i in range(0, len(data), batch_size):
        yield data[i : i + batch_size]

def extract_cls_embeddings(model, tokenizer, texts, device):
    encodings = tokenizer(
        texts,
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=128
    )
    input_ids = encodings["input_ids"].to(device)
    attention_mask = encodings["attention_mask"].to(device)
    with torch.no_grad():
        outputs = model(input_ids, attention_mask=attention_mask)
        cls_embeddings = outputs.last_hidden_state[:, 0, :]
    return cls_embeddings.cpu().numpy()

def introduce_gradual_drift(text_list, fraction_shuffle=0.5):
    new_texts = []
    for txt in text_list:
        words = txt.split()
        if len(words) < 2:
            new_texts.append(txt)
            continue
        k = int(len(words) * fraction_shuffle)
        if k < 1:
            new_texts.append(txt)
            continue
        indices = list(range(len(words)))
        random.shuffle(indices)
        shuffle_indices = indices[:k]
        to_shuffle = [words[i] for i in shuffle_indices]
        random.shuffle(to_shuffle)
        for i, idx in enumerate(shuffle_indices):
            words[idx] = to_shuffle[i]
        new_texts.append(" ".join(words))
    return new_texts

###############################################################################
# 2) Vector/Feature-Space Distance Functions
###############################################################################
def euclidean_distance(x, y):
    return np.sqrt(np.sum((x - y) ** 2))

def manhattan_distance(x, y):
    return np.sum(np.abs(x - y))

def minkowski_distance(x, y, p=3):
    return np.sum(np.abs(x - y) ** p) ** (1.0 / p)

def chebyshev_distance(x, y):
    return np.max(np.abs(x - y))

def canberra_distance(x, y):
    numerator = np.abs(x - y)
    denominator = np.abs(x) + np.abs(y) + 1e-12
    return np.sum(numerator / denominator)

DISTANCE_FUNCTIONS = {
    "euclidean": euclidean_distance,
    "manhattan": manhattan_distance,
    "minkowski": lambda u, v: minkowski_distance(u, v, p=3),  # example p=3
    "chebyshev": chebyshev_distance,
    "canberra": canberra_distance,
}

###############################################################################
# 3) EmbeddingTracker for Mahalanobis or the simpler distances
###############################################################################
class EmbeddingTracker:
    """
    A tracker that maintains:
    - An exponential-moving-average (EMA) of the embedding mean
    - An exponential-moving-average covariance (for Mahalanobis)
    - A selected distance metric: from one of the DISTANCE_FUNCTIONS
    """
    def __init__(self, embedding_dim, alpha=0.01, distance_name="mahalanobis"):
        self.alpha = alpha
        self.distance_name = distance_name

        # For all trackers, keep track of the mean.
        self.mean = np.zeros((embedding_dim,))
        self.count = 0

        # Mahalanobis-specific
        self.cov = np.eye(embedding_dim)

    def update(self, embedding):
        """Update the running mean (and covariance, if Mahalanobis) and cache the inverse covariance periodically."""
        update_interval = 10  # Update the cached inverse every 10 updates

        if self.count == 0:
            self.mean = embedding
            if self.distance_name == "mahalanobis":
                self.cov = np.eye(len(embedding))
                self.cov_inv = np.eye(len(embedding))  # initialize cached inverse covariance
        else:
            # Update the exponential moving average for the mean.
            self.mean = (1 - self.alpha) * self.mean + self.alpha * embedding

            if self.distance_name == "mahalanobis":
                diff = embedding - self.mean
                # Update the covariance using an exponential moving average.
                self.cov = (1 - self.alpha) * self.cov + self.alpha * np.outer(diff, diff)
                # Only update the inverse covariance periodically.
                if self.count % update_interval == 0:
                    try:
                        self.cov_inv = np.linalg.inv(self.cov)
                    except np.linalg.LinAlgError:
                        self.cov_inv = np.linalg.pinv(self.cov)
        self.count += 1


    def compute_distance(self, embedding, use_diagonal=True):
        if self.distance_name == "mahalanobis":
            diff = embedding - self.mean
            if use_diagonal:
                # Avoid division by zero with a small constant
                diag = np.diag(self.cov) + 1e-12
                cov_inv_diag = 1.0 / diag
                return np.sqrt(np.sum(diff**2 * cov_inv_diag))
            else:
                cov_inv = np.linalg.pinv(self.cov)
                return np.sqrt(diff.T @ cov_inv @ diff)
        else:
            dist_fn = DISTANCE_FUNCTIONS[self.distance_name]
            return dist_fn(self.mean, embedding)



def collect_data_single_seed(seed):
    set_seed(seed)

    # Detect device just once per seed
    if torch.backends.mps.is_available():
        device = torch.device("mps")
    elif torch.cuda.is_available():
        device = torch.device("cuda")
    else:
        device = torch.device("cpu")
    print(f"[Seed={seed}] Using device:", device)

    results = defaultdict(list)

    for dataset_info in args["datasets"]:
        dataset_name = dataset_info["name"]
        dataset_config = dataset_info["config"]
        dataset_split = dataset_info["split"]
        text_col = dataset_info["text_column"]

        ds = load_dataset(dataset_name, dataset_config, split=dataset_split)
        texts = list(ds[text_col])
        random.shuffle(texts)
        if args["max_texts"] > 0 and len(texts) > args["max_texts"]:
            texts = texts[: args["max_texts"]]

        # Split half for baseline, half for drift
        half_point = len(texts) // 2
        baseline_texts = texts[:half_point]
        drift_texts = texts[half_point:]

        for model_name in args["models"]:
            print(f"[Seed={seed}] --- Using Model: {model_name} ---")
            tokenizer = AutoTokenizer.from_pretrained(model_name)
            model = AutoModel.from_pretrained(model_name).to(device)
            model.eval()

            # Ensure gradients are off for speed
            with torch.no_grad():
                # ---- 1) Compute baseline embeddings ----
                baseline_emb_batches = []
                for b in batch_generator(baseline_texts, args["batch_size"]):
                    emb_b = extract_cls_embeddings(model, tokenizer, b, device)
                    baseline_emb_batches.append(emb_b)
                baseline_embs = np.concatenate(baseline_emb_batches, axis=0)

                # ---- 2) Compute baseline mean ----
                baseline_mean = baseline_embs.mean(axis=0)
                key = (dataset_name, model_name)

                # ---- 3) Fit dimension reducers on baseline ----
                kll_reducer = KLLDimensionReducer()
                kll_reducer.fit(baseline_embs)
                baseline_mean_2D_kll = kll_reducer.transform_one(baseline_mean)

                # Fit PCA (optionally use randomized solver for speed)
                pca = PCA(n_components=2, svd_solver='randomized')
                pca.fit(baseline_embs)
                baseline_mean_2D_pca = pca.transform(baseline_mean.reshape(1, -1)).flatten()

                # ---- 4) Create drifted texts once per drift_strength ----
                drifted_texts_by_strength = {
                    drift_strength: introduce_gradual_drift(
                        drift_texts, fraction_shuffle=drift_strength
                    )
                    for drift_strength in args["drift_strengths"]
                }

                # We'll need to loop through each drift_strength, but we only want
                # to run the forward pass once for each final set of test texts.

                # Precompute embeddings for each drift_strength so we don’t
                # re-run the model forward pass multiple times:
                test_embeddings_by_strength = dict()
                for drift_strength in args["drift_strengths"]:
                    test_texts = baseline_texts + drifted_texts_by_strength[drift_strength]
                    # Collect embeddings for each batch, then store them
                    test_emb_list = []
                    for batch in batch_generator(test_texts, args["batch_size"]):
                        test_emb_list.append(extract_cls_embeddings(model, tokenizer, batch, device))
                    # test_emb_list is a list of (batch_size, dim) arrays
                    test_embeddings_by_strength[drift_strength] = test_emb_list

            # Now do the distance computations (still no_grad, though it matters less here)
            all_distance_names = ["mahalanobis"] + list(DISTANCE_FUNCTIONS.keys())
            for distance_name in all_distance_names:
                for drift_strength in args["drift_strengths"]:
                    # Rebuild the same test_texts to match the above embeddings
                    test_texts = baseline_texts + drifted_texts_by_strength[drift_strength]
                    test_emb_list = test_embeddings_by_strength[drift_strength]

                    # === No PCA Tracker ===
                    tracker_no_pca = EmbeddingTracker(
                        baseline_embs.shape[1], alpha=0.01, distance_name=distance_name
                    )
                    tracker_no_pca.update(baseline_mean)

                    distance_scores_no_pca = []
                    time_overhead_no_pca = []
                    start_time_no_pca = time.time()

                    for emb_batch in tqdm(test_emb_list, leave=False):
                        # We used to call extract_cls_embeddings here, but we already have emb_batch
                        mean_emb = emb_batch.mean(axis=0)

                        overhead_start = time.time()
                        dist = tracker_no_pca.compute_distance(mean_emb)
                        overhead_end = time.time()

                        distance_scores_no_pca.append(dist)
                        time_overhead_no_pca.append(overhead_end - overhead_start)
                        tracker_no_pca.update(mean_emb)

                    end_time_no_pca = time.time()
                    results[key].append({
                        "distance_name": distance_name,
                        "drift_strength": drift_strength,
                        "pca": False,
                        "final_similarity": distance_scores_no_pca[-1] if distance_scores_no_pca else 0.0,
                        "time_taken": end_time_no_pca - start_time_no_pca,
                        "avg_overhead": np.mean(time_overhead_no_pca),
                        "seed": seed,
                    })

                    # === KLL 2D Tracker ===
                    tracker_kll = EmbeddingTracker(2, alpha=0.01, distance_name=distance_name)
                    tracker_kll.update(baseline_mean_2D_kll)

                    distance_scores_kll = []
                    time_overhead_kll = []
                    start_time_kll = time.time()

                    for emb_batch in tqdm(test_emb_list, leave=False):
                        emb_2d = kll_reducer.transform_batch(emb_batch)
                        mean_emb_2d = emb_2d.mean(axis=0)

                        overhead_start = time.time()
                        dist_kll = tracker_kll.compute_distance(mean_emb_2d)
                        overhead_end = time.time()

                        distance_scores_kll.append(dist_kll)
                        time_overhead_kll.append(overhead_end - overhead_start)
                        tracker_kll.update(mean_emb_2d)

                    end_time_kll = time.time()
                    results[key].append({
                        "distance_name": distance_name,
                        "drift_strength": drift_strength,
                        "pca": "kll_2d",
                        "final_similarity": distance_scores_kll[-1] if distance_scores_kll else 0.0,
                        "time_taken": end_time_kll - start_time_kll,
                        "avg_overhead": np.mean(time_overhead_kll),
                        "seed": seed,
                    })

                    # === PCA 2D Tracker ===
                    tracker_pca = EmbeddingTracker(2, alpha=0.01, distance_name=distance_name)
                    tracker_pca.update(baseline_mean_2D_pca)

                    distance_scores_pca = []
                    time_overhead_pca = []
                    start_time_pca = time.time()

                    for emb_batch in tqdm(test_emb_list, leave=False):
                        emb_2d_pca = pca.transform(emb_batch)  # 2D PCA space
                        mean_emb_2d_pca = emb_2d_pca.mean(axis=0)

                        overhead_start = time.time()
                        dist_pca = tracker_pca.compute_distance(mean_emb_2d_pca)
                        overhead_end = time.time()

                        distance_scores_pca.append(dist_pca)
                        time_overhead_pca.append(overhead_end - overhead_start)
                        tracker_pca.update(mean_emb_2d_pca)

                    end_time_pca = time.time()
                    results[key].append({
                        "distance_name": distance_name,
                        "drift_strength": drift_strength,
                        "pca": "pca_2d",
                        "final_similarity": distance_scores_pca[-1] if distance_scores_pca else 0.0,
                        "time_taken": end_time_pca - start_time_pca,
                        "avg_overhead": np.mean(time_overhead_pca),
                        "seed": seed,
                    })

    return results


def collect_data_multiple_seeds():
    all_results = defaultdict(list)
    for seed in range(args["num_seeds"]):
        seed_results = collect_data_single_seed(seed)
        for key, records in seed_results.items():
            all_results[key].extend(records)
    print("\nAll seeds complete!")
    return all_results


all_results = collect_data_multiple_seeds()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import math
from collections import defaultdict

def plot_distance_with_error_bars(all_results):
    """
    Plot error bars for No PCA, KLL 2D, and PCA 2D distance metrics across drift strengths.
    """
    # Aggregate: key = (dataset_name, model_name, distance_name, drift_strength, pca_flag)
    aggregator = defaultdict(list)
    for (dataset_name, model_name), records in all_results.items():
        for r in records:
            dist_name = r["distance_name"]
            drift_strength = r["drift_strength"]
            final_dist = r["final_similarity"]
            pca_flag = r["pca"]  # No PCA, kll_2d, or pca_2d

            key = (dataset_name, model_name, dist_name, drift_strength, pca_flag)
            aggregator[key].append(final_dist)

    # Compute mean and std for each key.
    aggregated_data = []
    for key, dist_values in aggregator.items():
        dataset_name, model_name, dist_name, drift_strength, pca_flag = key
        mean_val = np.mean(dist_values)
        std_val = np.std(dist_values)
        aggregated_data.append({
            "dataset_name": dataset_name,
            "model_name": model_name,
            "distance_name": dist_name,
            "drift_strength": drift_strength,
            "pca_flag": pca_flag,
            "mean_val": mean_val,
            "std_val": std_val
        })

    # Group the aggregated data by (dataset_name, model_name) for plotting.
    grouped = defaultdict(list)
    for row in aggregated_data:
        key = (row["dataset_name"], row["model_name"])
        grouped[key].append(row)

    for (dataset_name, model_name), rows in grouped.items():
        # Gather all unique distance names.
        dist_names = sorted(set(r["distance_name"] for r in rows))
        n_dists = len(dist_names)
        n_cols = 3
        n_rows = math.ceil(n_dists / n_cols)
        fig, axs = plt.subplots(n_rows, n_cols, figsize=(5 * n_cols, 4 * n_rows), sharex=True)
        axs = axs.flatten()
        fig.suptitle(f"{dataset_name} | {model_name}", fontsize=16)

        for i, dist_name in enumerate(dist_names):
            ax = axs[i]
            dist_rows = [r for r in rows if r["distance_name"] == dist_name]

            # Separate No PCA, KLL, and PCA results
            dist_rows_no_pca = sorted([r for r in dist_rows if r["pca_flag"] == False], key=lambda x: x["drift_strength"])
            dist_rows_kll = sorted([r for r in dist_rows if r["pca_flag"] == "kll_2d"], key=lambda x: x["drift_strength"])
            dist_rows_pca = sorted([r for r in dist_rows if r["pca_flag"] == "pca_2d"], key=lambda x: x["drift_strength"])

            x_no_pca = [r["drift_strength"] for r in dist_rows_no_pca]
            y_no_pca = [r["mean_val"] for r in dist_rows_no_pca]
            e_no_pca = [r["std_val"] for r in dist_rows_no_pca]

            x_kll = [r["drift_strength"] for r in dist_rows_kll]
            y_kll = [r["mean_val"] for r in dist_rows_kll]
            e_kll = [r["std_val"] for r in dist_rows_kll]

            x_pca = [r["drift_strength"] for r in dist_rows_pca]
            y_pca = [r["mean_val"] for r in dist_rows_pca]
            e_pca = [r["std_val"] for r in dist_rows_pca]

            # Plot error bars
            ax.errorbar(x_no_pca, y_no_pca, yerr=e_no_pca, fmt='-o', capsize=3, label="Regular")
            ax.errorbar(x_kll, y_kll, yerr=e_kll, fmt='-s', capsize=3, label="KLL-Sketch")
            ax.errorbar(x_pca, y_pca, yerr=e_pca, fmt='-^', capsize=3, label="PCA")
            ax.set_title(dist_name)
            ax.set_xlabel("Drift Strength")
            ax.set_ylabel("Distance (Mean ± Std)")
            ax.grid(True, linestyle='--', alpha=0.5)
            ax.legend()

        # Hide any unused subplots.
        for j in range(i+1, n_rows*n_cols):
            axs[j].set_visible(False)

        plt.tight_layout(rect=[0, 0, 1, 0.96])
        plt.show()

# Generate plots
plot_distance_with_error_bars(all_results)
```

```python
def plot_overhead_bar_aggregated(summary_overhead):
    """
    Create a grouped bar chart of overhead for No PCA, KLL, and PCA, aggregated across all
    datasets, models, and drift strengths, using a log scale.
    """
    grouped = (
        summary_overhead
        .groupby(["distance_name", "pca"])
        .agg(
            overhead_mean=("mean_overhead", "mean"),
            overhead_std=("mean_overhead", "std"),
        )
        .reset_index()
    )

    # Prepare data for grouped bar chart
    distance_names = sorted(grouped["distance_name"].unique())
    x_positions = np.arange(len(distance_names))
    bar_width = 0.3  # Adjusted for three categories

    fig, ax = plt.subplots(figsize=(8, 5))

    for i, pca_flag in enumerate([False, "kll_2d", "pca_2d"]):
        subset = grouped[grouped["pca"] == pca_flag]
        label = "Regular" if pca_flag == False else ("KLL-Sketch" if pca_flag == "kll_2d" else "PCA")
        ax.bar(
            x_positions + i * bar_width,
            subset["overhead_mean"],
            yerr=subset["overhead_std"],
            width=bar_width,
            capsize=5,
            alpha=0.7,
            label=label
        )

    ax.set_yscale("log")
    ax.set_xticks(x_positions + bar_width)
    ax.set_xticklabels(distance_names, rotation=45, ha="right")
    ax.set_title("Average Overhead by Distance Metric (Log Scale)")
    ax.set_xlabel("Distance Metric")
    ax.set_ylabel("Overhead (seconds, log scale)")
    ax.legend()
    plt.tight_layout()
    plt.grid(axis="y", linestyle="--", alpha=0.5)
    plt.show()


# Flatten the all_results dictionary into a list of records
rows = []
for (dataset, model), records in all_results.items():
    for record in records:
        record['Dataset'] = dataset
        record['Model'] = model
        # Ensure 'pca' column exists
        record['pca'] = record.get('pca', 'No PCA')
        rows.append(record)

df = pd.DataFrame(rows)

# Check the structure of df before proceeding
print(df.head())  # Debugging: Inspect DataFrame structure

# Compute summary statistics.
summary_overhead = df.groupby(["Dataset", "Model", "distance_name", "drift_strength", "pca"]).agg(
    mean_overhead=("avg_overhead", "mean"),
    std_overhead=("avg_overhead", "std")
).reset_index()

# Check if "pca" exists in summary_overhead
print(summary_overhead.head())  # Debugging: Verify presence of 'pca' column

# Plot using log scale and save results to CSV.
output_csv_path = os.path.join("../data-analysis/data-test", "summary_overhead_vector-distance.csv")

# Run the plotting function
plot_overhead_bar_aggregated(summary_overhead)
```
